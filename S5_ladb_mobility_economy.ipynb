## 🧩Paso 4: Analizar y resumir datos de movilidad

Como el dataset de tráfico contiene **múltiples registros por ciudad**. En esta parte, calcularás los promedios anuales por ciudad para simplificar el análisis y obtener una visión más clara de las tendencias generales.

### 4.1 Calcular promedios de tráfico por ciudad

**🎯Objetivo:**
Obtener una vista consolidada del tráfico promedio por ciudad y año, para analizar patrones generales sin depender de datos diarios.

**Instrucciones**

- Agrupa los datos por `city`, `country` y `year`.
- Calcula el promedio **solo de las métricas de tráfico más relevantes**: como `jams_delay`, `traffic_index_live`, `jams_length_kms`, `jams_count`, `mins_delay`, y tiempos de viaje (`travel_time_live_per_10kms_mins` y `travel_time_hist_per_10kms_mins`).
- Guarda el resultado como `traffic_city_year_2024`, mantén las columnas como variables (no índices).


<details>
<summary>Haz clic para ver la pista</summary>
Usa ".agg()" para aplicar funciones de promedio. Al final, reinicia el índice para mantener las columnas de la agrupación como variables (no índices).

```python
# Calcular los  promedios de trafico por ciudad, país y año
traffic_city_year_2024 = (
    traffic_2024
    .groupby(['city', 'country', 'year'])
    .agg({
        'jams_delay': 'mean',
        'traffic_index_live': 'mean',
        'jams_length_kms': 'mean',
        'jams_count': 'mean',
        'mins_delay': 'mean',
        'travel_time_live_per_10kms_mins': 'mean',
        'travel_time_hist_per_10kms_mins': 'mean'
    })
    .reset_index()
) # tu código aqui

# Mostrar resultado
traffic_city_year_2024.head()
```

## 🧩Paso 5: Unir movilidad y economía

Combinar datasets te permite analizar cómo se relacionan los indicadores económicos con los de movilidad.

### 5.1 Unir tráfico (tabla principal) con indicadores económicos

**🎯Objetivo:**
Combinar la información de tráfico y economía en un solo DataFrame para analizar cómo las condiciones económicas se relacionan con la movilidad urbana.

**Instrucciones**
- Selecciona solo las **columnas relevantes** de cada dataset (por ejemplo, variables clave de tráfico y de economía).
- Usa `.copy()` al crear subconjuntos para evitar modificar el dataset original.
- Une ambos DataFrames y define como **claves de unión** a `city` y `year`.
- Mantén solo las ciudades y años presentes en ambos datasets.
- Guarda el resultado en una nueva variable llamada `merged` y muestra las primeras 5 filas.


<details>
<summary>Haz clic para ver la pista</summary>
Aplica una unión de tipo "inner" para mantener las ciudades y años presentes en ambos datasets.

```python
# Seleccionar columnas clave de tráfico y economía
left_cols = ['city','country','year','jams_delay','traffic_index_live',
             'jams_length_kms','jams_count','mins_delay',
             'travel_time_live_per_10kms_mins','travel_time_hist_per_10kms_mins']

right_cols = ['city','year','city_gdp_capita','unemployment_pct','pm25','population']

# Usar .copy() para crear los dos nuevos datasets reducidos
traffic_2024_small = traffic_city_year_2024[left_cols].copy()
eco_2024_small = eco_2024[right_cols].copy()

# Unir datasets
merged = traffic_2024_small.merge(eco_2024_small, on=['city', 'year'], how='inner')

# Mostrar las primeras 5 filas
# tu código aquí
merged.head(5)
```

# Resumen ejecutivo: Movilidad urbana y productividad económica en Latinoamérica (2024)

**Contexto y objetivo**

El análisis buscó establecer si existe relación entre la congestión vehicular (jams_delay, traffic_index_live) y la productividad económica de las ciudades (city_gdp_capita). El hallazgo central es que esa relación es débil y positiva (r = 0.28), lo que contradice la hipótesis simple de "más tráfico, menos productividad". En cambio, la variable que sí muestra una asociación clara con la congestión es el desempleo (unemployment_pct), con una correlación negativa moderada-fuerte (r = −0.69): las ciudades con mayor congestión tienden a tener menor desempleo, probablemente porque reflejan mayor actividad económica y densidad de empleos concentrados en zonas urbanas saturadas.

**Cobertura de datos**

El análisis cubre el año 2024, con 15 ciudades de 7 países (Brasil, Colombia, Perú, Argentina, Uruguay, México y Chile), integrando métricas de tráfico en tiempo real con indicadores económicos (PIB per cápita, desempleo, calidad del aire y población).

**Hallazgos iniciales**

Ciudad de México es un valor atípico extremo: presenta el jams_delay más alto (2,833, más de 3 desviaciones estándar sobre la media de 630) junto con el desempleo más bajo (3.2%) y un PIB per cápita elevado (21,111), consistente con el patrón de alta congestión = alta actividad económica. Sao Paulo, Bogotá y Lima le siguen con niveles altos de congestión (1,729, 1,142 y 1,052 respectivamente). En el extremo opuesto, Montevideo y Fortaleza muestran la menor congestión y niveles de PIB per cápita muy dispares (26,176 vs. 8,512), confirmando que el tráfico por sí solo no predice el nivel de riqueza de una ciudad.

**Recomendaciones**

Al comparar específicamente Bogotá, Lima y Buenos Aires, Bogotá es la ciudad prioritaria para inversión en infraestructura de transporte: combina la congestión más alta de las tres (jams_delay = 1,142 frente a 1,052 de Lima y 571 de Buenos Aires), el PIB per cápita más bajo (11,442 frente a 13,472 de Lima y 18,117 de Buenos Aires) y el desempleo más alto (10.0% frente a 6.5% y 7.2%). Esta combinación —alta congestión, baja productividad y alto desempleo— sugiere que la infraestructura vial deficiente podría estar limitando el crecimiento económico en lugar de ser solo un síntoma de él. Se recomienda: (1) priorizar a Bogotá para un estudio de impacto económico de la congestión, (2) dado que la correlación general PIB–tráfico es débil, evitar decisiones de inversión basadas únicamente en esta relación y complementar con un modelo multivariable que incluya desempleo y población, y (3) monitorear a Ciudad de México como caso atípico que podría distorsionar comparaciones agregadas si se incluye en análisis regionales sin ajustar por outliers.
